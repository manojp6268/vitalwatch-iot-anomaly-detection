```python
import wfdb
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt

import warnings
warnings.filterwarnings('ignore')
```

## Raw ECG Signal Exploration
Dataset: MIT-BIH Arrhythmia Database | Patient: 100 | Age: 69M
Sampling Rate: 360 Hz | Duration: ~30 minutes | Lead: MLII

```python
# Streaming record 100 directly from PhysioNet - no manual download needed
record = wfdb.rdrecord('100', pn_dir='mitdb')

# Load the annotations (these mark where heartbeats occur)
annotation = wfdb.rdann('100', 'atr', pn_dir='mitdb')

print(record.__dict__)
```

```python
# Extract the ECG signal - channel 0 is MLII (standard ECG lead)
ecg_signal = record.p_signal[:, 0]

# Sampling frequency - how many data points per second
fs = record.fs  

# Plot 10 seconds of the signal
plt.figure(figsize=(14, 4))
plt.plot(ecg_signal[:10*fs], color='royalblue', linewidth=0.8)
plt.title('MIT-BIH Patient 100 - Raw ECG Signal (First 10 Seconds)')
plt.xlabel('Sample')
plt.ylabel('Amplitude (mV)')
plt.grid(True)
plt.tight_layout()
plt.show()

print(f"Sampling Rate: {fs} Hz")
print(f"Total samples: {len(ecg_signal)}")
print(f"Signal duration: {len(ecg_signal)/fs/60:.2f} minutes")
```

```python
# Get annotation samples and symbols
ann_samples = annotation.sample
ann_symbols = annotation.symbol

# Plot ECG with annotation markers
plt.figure(figsize=(14, 4))
plt.plot(ecg_signal[:10*fs], color='royalblue', linewidth=0.8, label='ECG Signal')

# Mark only annotations that fall within our 10 second window
for i, sample in enumerate(ann_samples):
    if sample < 10 * fs:
        plt.axvline(x=sample, color='red', alpha=0.5, linewidth=0.8)
        plt.text(sample, 1.5, ann_symbols[i], fontsize=7, color='red', ha='center')

plt.title('MIT-BIH Patient 100 - ECG with Beat Annotations (First 10 Seconds)')
plt.xlabel('Sample')
plt.ylabel('Amplitude (mV)')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Count how many of each beat type exists in the full record
ann_df = pd.DataFrame({
    'sample': ann_samples,
    'symbol': ann_symbols
})

print("Total annotations:", len(ann_df))
print("Beat type distribution:")
print(ann_df['symbol'].value_counts())
```

#### Out of 2274 beats, only 35 are abnormal. 
That's ~1.5%. This is a huge Class Imbalance and it surely is one of the core problem in healthcare.
